```python
EN_STOPWORDS = set(stopwords.words('english'))
```

### Global Configuration (DIKUNCI)

```python
# Reproducibility
RANDOM_STATE = 1
np.random.seed(RANDOM_STATE)

# TF-IDF Configuration
MAX_FEATURES = 3000
NGRAM_RANGE = (1, 2)
MIN_DF = 3
MAX_DF = 0.85

# Recommendation Config
TOP_N_RECOMMENDATION = 5
```

### Utility Function (Dipakai Semua Section)

```python
def clean_text(text):

    if pd.isna(text):
        return "unknown"

    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in EN_STOPWORDS]

    return " ".join(tokens)
```

# SECTION 3 — DATA CLEANING & PREPROCESSING

```python
CLEAN_INPUT_PATH = f"{DATA_DIR}/initial_clean.csv"

df = pd.read_csv(CLEAN_INPUT_PATH)

print("Data loaded for preprocessing")
print("Shape:", df.shape)
```

### Handle Missing Values

```python
TEXT_COLUMNS = ["posTitle", "mbrTitle", "companyName", "country"]
NUMERIC_COLUMNS = [
    "companyStaffCount",
    "connectionsCount",
    "followersCount",
    "avgMemberPosDuration",
    "avgCompanyPosDuration"
]
DATE_COLUMNS = ["startDate", "endDate"]

# Text
for col in TEXT_COLUMNS:
    df[col] = df[col].fillna("unknown")

# Numeric
for col in NUMERIC_COLUMNS:
    df[col] = df[col].fillna(df[col].median())

# Date
for col in DATE_COLUMNS:
    df[col] = df[col].fillna("unknown")

print("Missing values handled.")
```

### Text Normalization

```python
nltk.download('punkt_tab')
for col in TEXT_COLUMNS:
    df[col] = df[col].apply(clean_text)

print("Text normalization completed.")
```

### Convert Numeric Columns

```python
for col in NUMERIC_COLUMNS:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

df[NUMERIC_COLUMNS].describe()
```

### Date Cleaning & Validation

```python
def validate_date(value):
    if value == "unknown":
        return value
    try:
        pd.to_datetime(value)
        return value
    except:
        return "unknown"

for col in DATE_COLUMNS:
    df[col] = df[col].apply(validate_date)

print("Date validation completed.")
```

### Duplicate Handling

```python
before = df.shape[0]
df.drop_duplicates(
    subset=["posTitle", "companyName", "startDate"],
    inplace=True
)
after = df.shape[0]

print(f"Duplicates removed: {before - after}")
print("New shape:", df.shape)
```

# SECTION 5 — VECTORIZATION

### Load Engineered Dataset

```python
VECTORIZATION_INPUT_PATH = f"{DATA_DIR}/engineered_linkedin.csv"

df = pd.read_csv(VECTORIZATION_INPUT_PATH)

print("Data loaded for vectorization")
print("Shape:", df.shape)
```

### Separate Text & Numeric Features

```python
TEXT_FEATURE = "career_profile"
NUMERIC_FEATURES = [
    "companyStaffCount",
    "connectionsCount",
    "followersCount",
    "avgMemberPosDuration",
    "avgCompanyPosDuration"
]

X_text = df[TEXT_FEATURE].astype(str)
X_numeric = df[NUMERIC_FEATURES].astype(float)

print("Text shape   :", X_text.shape)
print("Numeric shape:", X_numeric.shape)
```

### TF-IDF Vectorization

```python
tfidf_vectorizer = TfidfVectorizer(
    max_features=MAX_FEATURES,
    ngram_range=NGRAM_RANGE,
    min_df=MIN_DF,
    max_df=MAX_DF
)

X_text_tfidf = tfidf_vectorizer.fit_transform(X_text)

print("TF-IDF shape:", X_text_tfidf.shape)
```

### Scale Numeric Features

```python
numeric_scaler = MinMaxScaler()

X_numeric_scaled = numeric_scaler.fit_transform(X_numeric)

print("Scaled numeric shape:", X_numeric_scaled.shape)
```

### Combine Text & Numeric Features

```python
X_numeric_sparse = sp.csr_matrix(X_numeric_scaled)

X_final = sp.hstack([X_text_tfidf, X_numeric_sparse])

print("Final feature matrix shape:", X_final.shape)
print("Matrix type:", type(X_final))
```

### Save Vectorizer, Scaler & Feature Matrix

```python
joblib.dump(tfidf_vectorizer, f"{MODEL_DIR}/tfidf_vectorizer.pkl")
joblib.dump(numeric_scaler, f"{MODEL_DIR}/numeric_scaler.pkl")
joblib.dump(X_final, f"{MODEL_DIR}/feature_matrix.pkl")

print("Vectorizer, scaler, and feature matrix saved.")
```

### Save Metadata for Recommendation Output

```python
metadata = df[[
    "career_step",
    "startDateParsed",
    "endDateParsed"
]].copy()

joblib.dump(metadata, f"{MODEL_DIR}/metadata.pkl")

print("Metadata saved.")
```

# SECTION 6 — RECOMMENDATION ENGINE

### Load Model Artifacts

```python
tfidf_vectorizer = joblib.load(f"{MODEL_DIR}/tfidf_vectorizer.pkl")
numeric_scaler   = joblib.load(f"{MODEL_DIR}/numeric_scaler.pkl")
X_final          = joblib.load(f"{MODEL_DIR}/feature_matrix.pkl")
metadata         = joblib.load(f"{MODEL_DIR}/metadata.pkl")

print("Model artifacts loaded")
print("Feature matrix shape:", X_final.shape)
```

### Helper: Vectorize New Profile

```python
def vectorize_profile(text_profile, numeric_features):

    # Text vector
    text_vec = tfidf_vectorizer.transform([text_profile])

    # Numeric vector
    numeric_arr = np.array(numeric_features).reshape(1, -1)
    numeric_vec = numeric_scaler.transform(numeric_arr)
    numeric_vec = sp.csr_matrix(numeric_vec)

    # Combine
    return sp.hstack([text_vec, numeric_vec])
```

### Recommendation: Similar Position

```python
def recommend_position(
    text_profile,
    numeric_features,
    top_n=TOP_N_RECOMMENDATION
):

    user_vector = vectorize_profile(text_profile, numeric_features)

    similarity_scores = cosine_similarity(user_vector, X_final).flatten()

    top_indices = similarity_scores.argsort()[::-1][:top_n]

    results = []
    for idx in top_indices:
        results.append({
            "career_step": metadata.iloc[idx]["career_step"],
            "similarity_score": round(float(similarity_scores[idx]), 4)
        })

    return results
```

### Recommendation: Career Path

```python
def recommend_career_path(
    text_profile,
    numeric_features,
    max_steps=5
):
    user_vector = vectorize_profile(text_profile, numeric_features)

    similarity_scores = cosine_similarity(user_vector, X_final).flatten()
    best_idx = similarity_scores.argmax()

    reference_step = metadata.iloc[best_idx]["career_step"]

    career_path = (
        metadata[metadata["career_step"] == reference_step]
        .sort_values("startDateParsed")
        .head(max_steps)
    )

    return career_path[["career_step", "startDateParsed", "endDateParsed"]]
```

### Test Recommendation

```python
sample_text = df.iloc[0]["career_profile"]
sample_numeric = df.iloc[0][NUMERIC_FEATURES].values

print("=== POSITION RECOMMENDATION ===")
recommend_position(sample_text, sample_numeric)

print("\n=== CAREER PATH ===")
recommend_career_path(sample_text, sample_numeric)
```
